```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
```

```python
h_plank = 1
omega = 1
k = 1
T = 1
gamma = 0.1
N = 21

# Временной интервал
t_span = (0, 100)
t_eval = np.linspace(0, 100, 2000)
```

```python
n_th = 1/(np.exp((h_plank * omega)/(k * T))-1)
```

```python
rho_0 = np.zeros((N,N))
rho_0[19,19] = 1
```

```python
print(rho_0)
```

```python
#psi = 1/np.sqrt(2) * (ket_19 + ket_20)
#psi = sum_i {C_i * ket_i}
#pho_ij = C_i * C*_j
pho_1920 = pho_1919 = pho_2019 = pho_2020 = 1/2
pho_0 = np.zeros((21,21))
pho_0[18,18] = pho_0[18,19] = pho_0[19,18] = pho_0[19,19] = 1/2
print(pho_0)
```

```python
def drho_dt(t, rho_f):
    global omega, gamma, n_th
    N = int(np.sqrt(len(rho_f)))
    rho = rho_f.reshape(N,N).astype(complex)
    drhodt = np.zeros((N,N), dtype=complex)

    for i in range(N):
      for j in range(N):
        slug1 =  - rho[i,j] * (-1j * omega * (i-j) + gamma/2 * (i+j) * (n_th + 1) + gamma/2 * n_th *(i+j+2))

        slug2, slug3 = 0, 0
        if i + 1 < N and j + 1 < N:
          slug2 = rho[i+1,j+1] * gamma * (n_th + 1) * np.sqrt(i+1) * np.sqrt(j+1)
        if i - 1 >= 0 and j - 1 >= 0:
          slug3 = rho[i-1,j-1] * gamma * n_th * np.sqrt(i) * np.sqrt(j)

        drhodt[i,j] = slug1 + slug2 + slug3

    return drhodt.flatten()
```

```python
rho_0f = rho_0.flatten()
pho_0f = rho_0.flatten()
```

```python
# Решение ОДУ
sol = solve_ivp(lambda t, y: drho_dt(t, y),
                t_span, rho_0f, t_eval=t_eval, method='RK45')
```

```python
print(sol)
```

```python
# Извлечение диагональных элементов ρ_ii(t)
populations = []
for i in range(N):
    idx = i * N + i           # позиция элемента (i,i) в плоском векторе
    pop_i = sol.y[idx, :]
    populations.append(np.real(pop_i))
I = np.linspace(0, N, 2000)
```

```python
# Построение графика
plt.figure(figsize=(8, 5))
for i in range(min(5, N)):
    plt.plot(I, populations[i], label=f'ρ_{i}{i}')
plt.xlabel('i')
plt.ylabel('rho_ii')
plt.legend()
plt.grid(True)
plt.show()
```
